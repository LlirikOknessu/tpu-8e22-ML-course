```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
data=pd.read_csv("redacted_data.csv")

target = data["charges"]
features = data.drop(columns="charges")

x_train, x_test, y_train, y_test=train_test_split(features,
                                                  target,
                                                  train_size=0.8,
                                                  test_size=0.2,
                                                  random_state=10
)

features.head()
```

```python
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
data = x_test
data["charges_pred"]=model.predict(x_test.to_numpy())
data["charges_real"]=y_test
data["error"] = data["charges_real"]-data["charges_pred"]

mse = np.sqrt(np.mean(data["error"]**2))
print(mse)
```
